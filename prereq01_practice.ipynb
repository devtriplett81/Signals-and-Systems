```python
import numpy as np


u = np.array([6.0, 7.0, 6.0])
v = np.array([7.0, 6.0, 7.0])
w = np.array([6.0, 7.0, 6.0])

print("u = ", u)
print("v = ", v)
print("w = ", w)

M = np.column_stack((u, v, w))
print("Matrix M: ")
print(M)

print("Rank:")
print(M)

rank_M = np.linalg.matrix_rank(M)


print("Rank:",rank_M)
print("Number of vectors:", M.shape[1])
print("Linearly independent?", rank_M == M.shape[1])
```
