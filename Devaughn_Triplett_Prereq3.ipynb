```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Circuit stats
R = 330
C = 1e-6

# Sinusoidal input
A = 5
omega = 2
T = 10

take_constant = R*C
# samples from 0 to T seconds
t = np.linspace(0, T, 1000)

print(f"Resistance: {R:.0f} ohms")
print(f"Capacitance: {C:.1e} F")
print(f"Time constant: {take_constant:.2e} s")
print(f"Input amplitude: {A:.1f} V")
print(f"Input frequency: {omega:.1f} rad/s")
print(f"Simulation time: {T:.1f} s")

def sinusoidal_input(t):
    "Voltage input: x(t) = A * sin(omega * t)"
    return A * np.sin(omega * t)
def rc_ode(y,t):
    "RC circuit DE: RC(dy/dt)+y(t) = -x(t)"
    "Solving for dy/dt: dy/dt = [-x(t) - y(t)] / (RC)"
    x_t = sinusoidal_input(t)
    dydt = (x_t - y[0] / (R*C))
    return dydt

# Intial capacitor voltage
y_initial = [0]

# Solvlve DE numerically
solution = odeint(rc_ode, y_initial, t)

# returns 2 dimensional array, extracts first colum
y = solution[:,0]

# Evaluate input at every simulation time

x = sinusoidal_input(t)

print(f"Initial input voltage: x(0) = {x[0]:.6f}V")
print(f"Initial output voltage: y(0) = {y[0]:.6f}V")
print(f"Final input voltage: x(T) = {x[-1]:.6f}V")
print(f"Final output voltage: y(T) = {y[-1]:.6f}V")

fig, axes = plt.subplots(1,2, figsze=(14,5))

# Input plot

axes[0].plot(
    t,
    x,
    color = "green",
    linewidth = 2,
    label = r"$x(t) = 5\sin(2t)$"
)

axes[0].set_title("Sinusodial input volt")
axes[0].set_xlabel("Time (s)")
axes[0].set_ylabel("Input voltage x(t) (V)")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1
    ].plot(
    t,
    y,
    color = "pink",
    linewidth = 2,
    label = r"$y(t)$"
)
axes[1].set_title("RC output voltage")
axes[1].set_xlabel("Time (s)")
axes[1].set_ylabel("Output voltage y(t) (V)")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

fig.suptitle("RC circuit response to sinusoidal input",
             fontsize = 15
             fontweight = "bold"
)

plt.tight_layout()
plt.show()
    

```
