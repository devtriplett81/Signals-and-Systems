```python
%pip install matplotlib
import numpy as np
import matplotlib.pyplot as plt

# x1[n] = (1/3)^n u[n]

# plot values
n_plot = np.arange(-5,11)

# discrete-time unit step signal
x1_plot = np.zeros_like(n_plot, dtype=float)
nonnegative_plot = n_plot >= 0
x1_plot[nonnegative_plot] = (1/3)**n_plot[nonnegative_plot].astype(float)

# plot discrete time singal
plt.figure(figsize=(9,4.5))
plt.stem(n_plot, x1_plot, basefmt="k-")
plt.title(r"Problem 3a: $x_1[n]=(1/3)^n u[n]$")
plt.xlabel(r"Sample index $n$")
plt.ylabel(r"Amplitude $x_1[n]$")
plt.xticks(n_plot)
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.show()

# Confirm energy and power
N = 1000
n_test = np.arange(-N,N+1)
x1_test = np.zeros_like(n_test, dtype=float)
nonnegative_test = n_test >= 0
x1_test[nonnegative_test] = (1/3)**n_test[nonnegative_test].astype(float)

energy_x1 = np.sum(np.abs(x1_test)**2)
power_x1 = energy_x1 / (2 * N + 1)

print(f"Estimated energy over n = {-N} to {N}: {energy_x1:.8f}")
print(f"Exact energy: {9/8:.8f}")
print(f"Estimated average power: {power_x1:.8f}")
print("Expected average power as N approaches infinity: 0")

import numpy as np
import matplotlib.pyplot as plt

# x2(t) = cos(3t + pi/7)

# Values for plotting
t_plot = np.linspace(-2 * np.pi, 2 * np.pi, 3000)
x2_plot = np.cos(3 * t_plot + np.pi / 7)

# continuous-time signal
plt.figure(figsize=(10, 4.5))
plt.plot(t_plot, x2_plot, linewidth=2)
plt.axhline(0, color="black", linewidth=0.8)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.8)
plt.axhline(-1, color="gray", linestyle="--", linewidth=0.8)

plt.title(r"Problem 3(b): $x_2(t)=\cos(3t+\pi/7)$")
plt.xlabel(r"Time $t$")
plt.ylabel(r"Amplitude $x_2(t)$")
plt.ylim(-1.2, 1.2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# confirm that energy increases without bound
# while average power approaches 1/2
dt = 0.001

print("Window half-width T | Energy on [-T,T] | Average power")

for T in [10, 50, 200]:
    t_test = np.arange(-T, T, dt)
    x2_test = np.cos(3 * t_test + np.pi / 7)

    energy_x2 = np.sum(np.abs(x2_test) ** 2) * dt
    power_x2 = energy_x2 / (2 * T)

    print(f"{T:19d} | {energy_x2:16.6f} | {power_x2:.6f}")

print("Energy keeps increasing as T increases, so E_infinity = infinity.")
print("Average power approaches 0.5.")

# x3[n] = sin(pi*n/3 + pi/10)

# Values for plotting
n_plot = np.arange(-6, 19)
x3_plot = np.sin(np.pi * n_plot / 3 + np.pi / 10)

# Plot discrete-time signal
plt.figure(figsize=(10, 4.5))
plt.stem(n_plot, x3_plot, basefmt="k-")
plt.axhline(1, color="gray", linestyle="--", linewidth=0.8)
plt.axhline(-1, color="gray", linestyle="--", linewidth=0.8)

plt.title(r"Problem 3(c): $x_3[n]=\sin(\pi n/3+\pi/10)$")
plt.xlabel(r"Sample index $n$")
plt.ylabel(r"Amplitude $x_3[n]$")
plt.xticks(n_plot)
plt.ylim(-1.2, 1.2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Confirm fundamental period N0 = 6
n_check = np.arange(-20, 21)

x3_now = np.sin(np.pi * n_check / 3 + np.pi / 10)
x3_shifted = np.sin(np.pi * (n_check + 6) / 3 + np.pi / 10)

period_six_verified = np.allclose(x3_now, x3_shifted)

# Calculate energy and power over one period
n_period = np.arange(0, 6)
x3_period = np.sin(np.pi * n_period / 3 + np.pi / 10)

energy_one_period = np.sum(np.abs(x3_period) ** 2)
power_x3 = energy_one_period / 6

print(f"Does x3[n + 6] equal x3[n]? {period_six_verified}")
print(f"Energy in one period: {energy_one_period:.8f}")
print(f"Average power: {power_x3:.8f}")

# Show that total energy increases without bound
print("\nNumber of periods | Energy over those periods")

for number_of_periods in [1, 5, 20, 100]:
    total_energy = number_of_periods * energy_one_period
    print(f"{number_of_periods:17d} | {total_energy:.6f}")

print("Energy grows without bound, so E_infinity = infinity.")
```
