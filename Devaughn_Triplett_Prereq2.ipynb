```python
import numpy as np

a = np.array([2,-1])
b = np.array([1,3])

za = a[0] + 1j*a[1]
zb = b[0] + 1j*b[1]

print("Vector a:",a)
print("Vector b:",b)

print("za = ",za)
print("zb = ",zb)

dot_product = np.dot(a,b)

complex_inner_product = np.conj(za)*zb
real_part = np.real(complex_inner_product)


print("\nReal vector dot product:",dot_product)
print("conj(za)*zb:",complex_inner_product)
print("Real:",real_part)
print("Dot products match? ;-;",np.isclose(dot_product,real_part))
print("Are the 2d vectors orthogonal? ",np.isclose(dot_product,0.0))


print("The real vector dot productand the real part of the complex inner part are equal -1")
print("Vectors are not orthogonal because the dot product is not zero")




```
