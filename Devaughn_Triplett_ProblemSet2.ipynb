```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from pathlib import Path

wav_path = Path("guitar_clean.wav")

if not wav_path.exists():
    raise FileNotFoundError(
        "guitar_note.wav not found. "
        "Place it in same folder as notebook"
    )

sample_rate, x_raw = wavfile.read(wav_path)

print("Sample rate: ", sample_rate, "hz")
print("Original data type:", x_raw.dtype)
print("Original array shape: ",x_raw.shape)

if np.issubdtype(x_raw.dtype, np.signedinteger):
    limits = np.iinfo(x_raw.dtype)
    scale = max(abs(limits.min), limits.max)
    x = x_raw.astype(np.float64) / scale

elif np.issubdtype(x_raw.dtype, np.unsignedinteger):
    limits = np.iinfo(x_raw.dtype)
    midpoint = (limits.max + 1) / 2
    x = (x_raw.astype(np.float64) - midpoint) / midpoint
else:
    x = x_raw.astype(np.float)

if x.ndim == 2:
    x = np.mean(x, axis = 1)
duration = len(x) / sample_rate

duration = len(x) / sample_rate

print("Processed array shape: ", x.shape)
print(f"Audio duration: {duration:.3f} seconds")
print(f"Original peak amp: {np.max(np.abs(x)):.6f}")


alpha_values = [1, 2, 5, 10]

start_time = 0.0
view_duration = 0.02  

start_sample = int(start_time * sample_rate)
end_sample = min(
    len(x),
    start_sample + int(view_duration * sample_rate)
)

x_slice = x[start_sample:end_sample]
time_slice = np.arange(start_sample, end_sample) / sample_rate

for alpha in alpha_values:
    amplified_slice = alpha * x_slice

    plt.figure(figsize=(11, 4.5))

    plt.plot(
        time_slice,
        x_slice,
        label=r"Original $x(t)$",
        linewidth=1.5
    )

    plt.plot(
        time_slice,
        amplified_slice,
        label=rf"Amplified $\alpha x(t)$, $\alpha={alpha}$",
        linewidth=1.2,
        alpha=0.8
    )

    plt.axhline(0, color="black", linewidth=0.8)

    plt.title(
        rf"Problem 3(a): Original and Amplified Guitar Signal, "
        rf"$\alpha={alpha}$"
    )
    plt.xlabel("Time (seconds)")
    plt.ylabel("Amplitude")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()

    
    plt.savefig(
        f"problem3a_alpha_{alpha}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```
