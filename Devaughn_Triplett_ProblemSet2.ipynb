```python
%pip install scipy
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from pathlib import Path

wav_path = Path("guitar_clean.wav")

if not wav_path.exists():
    raise FileNotFoundError(
        "guitar_note.wav not found. "
        "Place it in same folder as notebook"
    )

sample_rate, x_raw = wavfile.read(wav_path)

print("Sample rate: ", sample_rate, "hz")
print("Original data type:", x_raw.dtype)
print("Original array shape: ",x_raw.shape)

if np.issubdtype(x_raw.dtype, np.signedinteger):
    limits = np.iinfo(x_raw.dtype)
    scale = max(abs(limits.min), limits.max)
    x = x_raw.astype(np.float64) / scale

elif np.issubdtype(x_raw.dtype, np.unsignedinteger):
    limits = np.iinfo(x_raw.dtype)
    midpoint = (limits.max + 1) / 2
    x = (x_raw.astype(np.float64) - midpoint) / midpoint
else:
    x = x_raw.astype(np.float)

if x.ndim == 2:
    x = np.mean(x, axis = 1)
duration = len(x) / sample_rate

duration = len(x) / sample_rate

print("Processed array shape: ", x.shape)
print(f"Audio duration: {duration:.3f} seconds")
print(f"Original peak amp: {np.max(np.abs(x)):.6f}")
```
